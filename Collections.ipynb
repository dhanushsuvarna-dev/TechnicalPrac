```python
#List on Number
nums = [1,2,3,4,5]
#List of String
colors = ["Red","Black","Green"]
#List allow duplicate values
color = ["Red","Black","Green","Red"]
#List of Mixed datatype
#nested List
mat = [[1,2,3],[45,65,23],[89],[43,67]]
```

```python
fruits = ["apple","orange","cherry","apple"]
print(fruits)
```

```python
#indexing & silcing
print(fruits[2])
print(fruits[-3])
#Slicing ->string_name[start:stop:stop]
print(fruits[1:3])
rev_fruits = fruits[::-1]
print(rev_fruits)
```

```python
#Modifying list
print("Org:",fruits)
#add the element end of the list
fruits.append("Mango")
print(fruits)
#To find the length of the list
len_fruits = len(fruits)
print("Length: ",len_fruits)
#To remove the last element
fruits.pop()
print(fruits)
```

```python
nums = [4,2,5,3,6,1]
print(max(nums)) #Max Element
print(min(nums)) #Min Element
#Quick Sort
nums.sort()
print(nums)
```

```python
#Looping
student = ["Raju",24,7.23,True]
#Using indexing
for i in range(len(student)):
    print(student[i],end=" ")
print("\n")
#Use
for val in student:
    print(val,end=" ")
```
