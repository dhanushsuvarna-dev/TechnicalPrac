```python
import matplotlib.pyplot as plt
import pandas as pd
```

Pie Chart

```python
monthly_data = {
    "Brand" : ["Oneplus","Apple","Vivo","Redmi","Moto"],
    "Sales" : [16,5,19,23,17]
}
sales_df = pd.DataFrame(monthly_data)
sales_df

plt.figure(figsize=(8,6))
plt.pie(sales_df['Sales'],
        labels=sales_df['Brand'],
        colors=['g','r','b','m','orange'],
        autopct="%.1f%%",
        explode=[0,.1,0,0,0],
        shadow=True,
        startangle=180,
        pctdistance=0.7
        )
plt.title("Mobile Sales Distribution",fontsize=13,fontweight="bold")
plt.legend(sales_df["Brand"],title="Mobile Brands", loc='right')
plt.tight_layout()
plt.show()
```

```python
df = pd.read_excel("expense.xlsx")
cat=df.groupby('Payment Mode')["Amount"].sum()
plt.figure(figsize=(8,6))
plt.title("Payment Mode")
plt.pie(cat.values,
        labels=cat.index,
        explode=[0,0.05,0],
        autopct="%.2f%%")
plt.show()
```

Histogram

```python
weather = { 
    "Time" : ['9.00','9.30','10.00','10.30','11.00','11.30','12.00','12.30','13.00','13.30','14.00','14.30','15.00','15.30'],
    "temp" : [33,30,30,32,33,31,32,33,32,35,32,34,33,32]
}
weather_df = pd.DataFrame(weather)
plt.hist(
    weather_df['temp'],
    bins=6,
    color='skyblue',
    edgecolor = 'black'
)
plt.show()
```

```python
weekly_Sales = {
    "Day":['Monday','Tuesday','Wednesday','Thursday','Friday','Saturnday','Sunday'],
    "week1":[15,20,18,8,10,19,30],
    "week2":[24,14,10,18,13,17,35],
    "week3":[10,40,50,23,69,82,43]
}
sales_df = pd.DataFrame(weekly_Sales)
plt.stackplot(
    sales_df['Day'], sales_df['week1'], sales_df['week2'],sales_df['week3'],
    labels=["Week1","Week2","Week3"]
)
plt.legend()
plt.show()
```
