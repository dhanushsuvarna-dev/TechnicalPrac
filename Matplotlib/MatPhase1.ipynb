Matplotlib us a comprehensive library for creating statics, animated and interaction visualization in python .matplotlib

```python
import matplotlib.pyplot as plt
import pandas as pd
```

Bar Plot

```python
c_name = ["Java","Python","C","C++"]
enroll = [24,30,12,14]

plt.bar(c_name,enroll)
plt.show()
```

```python
coursedata={
    "cname":["Java","Python","c","C++"],
    "enroll":[24,30,12,15]
}
course_df=pd.DataFrame(coursedata)
bars = plt.bar(
    course_df["cname"],
    course_df["enroll"],
    color=["#ADE827","#2DCCB8","#CC2D52","#CC2DC9"],
    edgecolor = "black",
    width=0.5
)
#Add Title and label
plt.title("Course enroll Overview")
plt.ylabel("Enroll Students",fontsize=14,fontweight="bold")
plt.xlabel("Course Name",fontsize=12)
plt.bar_label(bars,padding=5,fontsize=12,fontweight="bold")
plt.ylim(0,50)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.grid(axis="y",linestyle="--",alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
data = pd.read_excel("expense.xlsx")
exp_df = pd.DataFrame(data)
expgroupby = exp_df.groupby("Category")['Amount'].sum()

bar = plt.bar(expgroupby.index,
        expgroupby.values,
        color=["#ADE827","#2DCCB8","#CC2D52","#CC2DC9"],
            edgecolor = "black",
            width=0.5
        )
plt.ylabel("Money in Rupees",fontsize=14,fontweight="bold")
plt.xlabel("Expenses",fontsize=14,fontweight="bold")
plt.bar_label(bar,padding=5,fontsize=12,fontweight="bold")
plt.ylim(0,5000)
plt.title("The Customer List",fontsize=14,fontweight="bold")
plt.show()
```

```python
weekly_Sales = {
    "Day":['Monday','Tuesday','Wednesday','Thursday','Friday','Saturnday','Sunday'],
    "week1":[15,20,18,8,10,19,30],
    "week2":[24,14,10,18,13,17,35],
}
sales_df = pd.DataFrame(weekly_Sales)
plt.plot(sales_df['Day'],sales_df['week1'],marker="o",ls="-.",label="Week1_Sales",color="blue")
plt.plot(sales_df['Day'],sales_df['week2'],marker="o",ls="--",label="Week2_Sales",color="green")
plt.title("Weekly Sales Overview",fontsize=21)
plt.xlabel("week1_sales")
plt.ylabel("week2_sales")
plt.legend()
plt.show()
```

```python
adv_sales={
    "Advertising":[10,20,25,25,25,20,29,35],
    "Sales":[130,280,300,230,340,400,250,360]
}
sales_df=pd.DataFrame(adv_sales)
plt.scatter(
    sales_df["Advertising"],
    sales_df["Sales"],
    color="red",
    marker="*",
    s=200,
    label="Sales vs Advertising"
)
plt.title("Advertising expenditure V/S Sales Revenue")
plt.xlabel("Advertising expenditure(in $ thousands)")
plt.show()
```
