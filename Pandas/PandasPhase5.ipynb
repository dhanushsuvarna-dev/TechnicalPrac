##  Compare,Pivoting & Melting DataFrame

```python
import pandas as pd
```

Compare

```python
shop_7Oct = {
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[150,200,60,25],
    "Quantity":[15,7,12,13]
}
Stock_7Oct= pd.DataFrame(shop_7Oct)

shop_8Oct = {
    "Fruits":['Mango','Apple','Banana','Papaya'],
    "Price":[200,250,60,35],
    "Quantity":[15,17,24,15]
}

Stock_8Oct=pd.DataFrame(shop_8Oct)
Stock_8Oct
```

```python
Stock_7Oct.compare(Stock_8Oct)
Stock_7Oct.compare(Stock_8Oct,align_axis=0)
Stock_7Oct.compare(Stock_8Oct,keep_equal=True)
Stock_7Oct.compare(Stock_8Oct,keep_shape=True)
Stock_7Oct.compare(Stock_8Oct,align_axis=0,keep_shape=True,keep_equal=True)
```

```python
monthly_shop = {
    "Product":['Laptop','Mobile','Laptop','Mobile','Laptop','Laptop'],
    "Month":["Jan","Jan","Feb","Feb","Mar","Apr"],
    "Price":[50000,27500,65000,45000,95000,35000]
}

monthly_sales = pd.DataFrame(monthly_shop)
monthly_sales
```

```python
monthly_sales.pivot(
    index = "Product",
    columns="Month",
    values="Price"
)
```

```python
monthly_sales.pivot_table(values="Price",index="Product",columns="Month",aggfunc="sum")
```

```python
shop_sales = {
    "product" : ['Laptop','Mobile','Mouse'],
    "Jan" : [67500,25000,950],
    "Feb" : [87000,35000,850],
    "Mar" : [98000,23500,650]
}
sales = pd.DataFrame(shop_sales)

sales.melt(
    id_vars = "product",
    var_name = "Month",
    value_name = "Sales"
)
```
