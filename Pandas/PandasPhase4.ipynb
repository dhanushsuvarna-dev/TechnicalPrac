### GroupBy & MergeorJoin

```python
import pandas as pd
```

```python
exp = pd.read_excel("Subdata_pandas/expense.xlsx")
exp
```

```python
#Find the total expense for each category
#M1
exp.groupby("Category")["Amount"].sum()
#M2
exp.groupby("Category")["Amount"].agg(['sum'])

#Give a complete expense summary for each category
exp.groupby("Category")["Amount"].agg(
    ['sum','count','mean','max','min']
)
```

```python
# How much money was paid using cash,card and Card?
exp.groupby("Payment Mode")["Amount"].agg(['sum'])
# How many transactions were made using each payment mode?
exp.groupby("Payment Mode")["Amount"].agg(['count'])
```

```python
#for each category how much amount spent using card,cash & UPI
exp.groupby(['Category','Payment Mode'])["Amount"].agg(['sum'])

#For each payment mode,calculate the total amount spent across categories
exp.groupby(['Payment Mode','Category'])["Amount"].agg(['sum'])
```

```python
#which categories have the highest total expense
exp.groupby('Category')['Amount'].agg('sum').sort_values(ascending=False)
#create a Dataframe containing the total expense of each category
exp.groupby('Category')['Amount'].sum().reset_index(name="Total Expense")
```

# Join OR Merge

```python
Customer = pd.read_csv("Subdata_pandas/customer.csv")
Customer
payment = pd.read_csv("Subdata_pandas/payment.csv")
payment
```

```python
#Inner Join
pd.merge(Customer,payment,on="customerid", how="inner")
```

```python
#Left Join
pd.merge(left=Customer,right=payment,on="customerid",how="left")
```

```python
#Right Join
pd.merge(left=Customer,right=payment,on="customerid",how="right")
```
