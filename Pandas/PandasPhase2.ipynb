### Dealing with Duplicate data and Missing Value

```python
import pandas as pd
import numpy as np
```

Duplicate Data

```python
cdata = pd.read_csv("Subdata_pandas/company.csv")
cdata
```

```python
cdata.duplicated()
#Check emp id
cdata['EEID'].duplicated()
cdata['Name'].duplicated()
#Drop Data
cdata['EEID'].drop_duplicates()
```

```python
#Find null values
cdata.isnull().sum()
#Element
cdata.dropna()
```

```python
#Replace Nan with any value
cdata.replace(np.nan, "EmptyData")
cdata['salary'].replace(np.nan,5000)
#Replace Nan in salary with mean value
mean = np.mean(cdata['salary'])
cdata['salary'].replace(np.nan, mean)
#Method 1
cdata['salary'] = cdata['salary'].fillna(mean)
cdata
```

```python
cdata = pd.read_csv("Subdata_pandas/company.csv")
cdata
#Forward Fill previous row
cdata.ffill()
#Backward Fill previous row
cdata.bfill()
```
