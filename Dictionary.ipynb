Dictionary Method

```python
student1 = {
    'id' : "Cs038",
    'name' : "Dhanush",
    'age' : 23,
    'dept' : "CSE",
    'cgpa' : 8.40
}
s_name = student1.get('name')
print(s_name)
k = student1.keys()
print(k)
v = student1.values()
print(v)
d = student1.items()
print(d)
c = student1.copy()
print(c)
#defualt
dvalue =  student1.setdefault("pin",100123)
print(dvalue)
print(student1)
#Update
student1.update({"dept":"CSEA"})
print(student1)
#pop
pop_val = student1.pop('cgpa')
print(pop_val)
print(student1)
#popitem
student1.popitem()
print(student1)
student1.clear()
print(student1)
```

```python
student = {
    'id' : "24cs049",
    'name':"Ganapathi",
    'age': 23
}
for key in student:
    print(key)
for xv in student.values():
    print(xv)
for k,v in student.items():
    print(k,":",v)
```
