Python Tuples

A tuples is a collection of items in Python that is ordered,unchangeable(immutable) and allow duplicate values


```python
#-------Types-------

#Type1: Using Parentheses
nums = {1,2,3,4}
mixed = ("Raju",24,43,56,True)
nested=(21,(23,54),(67,68,69))
print(nested,type(nested))
print(mixed,type(mixed))
print(nums,type(nums))#sets

#Type 2:Without using Parentheses
color1 = "red","green","blue"
print(color1,type(color1)) 

#Type 3: Using Tuple Const
list1=['a','b','c','d']
tuple_list=tuple(list1)
print(tuple_list, type(tuple_list))

#Type 4:Single
value=(50,)
print(value, type(value))
```

Tuple Operation

```python
#Concatenation
del list
tup2 = (10,20,30,40)
tup1 = (1,2,3,4)
tup3 = tup1+tup2;
print(tup3)
#Converting to List nd changing the tuple value
list_Tup = list(tup2)
list_Tup[3] = 35
tup2 = tuple(list_Tup)
print(tup2)
```

```python
marks = (40)*5
print(marks)
#Search
tup=(10,20,30,40)
print(50 in tup)
```

Iterating Over Tuples

```python
color = ("blue","black","green","yellow")
#for loop
for c in color :
    print(c,end=",")
#while
print("\n")
i=0
while i<len(color):
    print(color[i],end=" ")
    i+=1
```

Tuple Methods

```python
colors = ("blue","black","green","yellow","blue","red")
#count
print(colors.count("blue"))
#index
print(colors.index("red"))
```

```python
nums = (10,32,45,67,87,67,89,90)
#length
print(len(nums))
#max and min
print("max=",max(nums),"min=",min(nums))
#sort
sortedNums = sorted(nums)
print(sortedNums)
```

Packing & Unpacking

```python
v1 ="Raju"
v2 = 23
v3 = 9.76
S001 = v1,v2,v3

name,age,cgpa = S001
print(name)
print(age)
print(cgpa)
```
