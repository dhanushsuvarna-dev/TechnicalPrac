```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Loading Student File
student_df = pd.read_csv("student_scores.csv")
student_df.head(10)
```

```python
#Range in Math Score
max_r=student_df['MathScore'].max()
min_r=student_df['MathScore'].min()
max_r,min_r
```

```python
#Create Male DataSet
male_df = student_df[student_df['Gender']=='male']
male_df
#Create Female DataSet
female_df = student_df[student_df['Gender']=='female']
female_df
```

```python
#Range in WritingScore
max_r = male_df['WritingScore'].max()
min_r = male_df['WritingScore'].min()
range_Writing = max_r - min_r
max_r,min_r,range_Writing
```

```python
#Range in WritingScore
max_r = male_df['ReadingScore'].max()
min_r = male_df['ReadingScore'].min()
range_Writing = max_r - min_r
max_r,min_r,range_Writing
```

```python
IT_Department_A = {
    "Employee ID": ["E101", "E102", "E103", "E104", "E105", "E106"],
    "Employee Name": ["Rohan", "Kiran", "Aditya", "Sneha", "Pooja", "Rahul"],
    "Project": [75, 65, 73, 68, 72, 67]
}
IT_Department_B= {
    "Employee ID": ["E201", "E202", "E203", "E204", "E205", "E206"],
    "Employee Name": ["Arjun", "Deepak", "Kavya", "Megha", "Sanjay", "Varun"],
    "Project": [90, 47, 93, 47, 96, 51]
}

deptA_df = pd.DataFrame(IT_Department_A)
deptB_df = pd.DataFrame(IT_Department_B)
```

```python
plt.scatter([i for i in range(0,len(deptA_df["Project"]))],
            deptA_df["Project"],color='red',s=100,label="Dept-A")
plt.scatter([i for i in range(0,len(deptB_df["Project"]))],
            deptB_df["Project"],color='green',s=100,label="Dept-B")
meanVal = deptA_df['Project'].mean()
plt.axhline(y=meanVal,color='orange',linestyle='--',label='Mean')
plt.legend()
plt.show()
```
