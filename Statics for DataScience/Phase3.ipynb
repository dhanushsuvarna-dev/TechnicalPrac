### Percentile
Percentile are used in statics to give you a number that describe that value that given percent of the values are lower than

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic_df = pd.read_csv('titanic.csv')
```

```python
titanic_df.describe()
```

```python
titanic_df.isnull().sum()
```

```python
#fill all null values on Age col with mean
mean_age = titanic_df['Age'].mean()
titanic_df['Age'] =  titanic_df['Age'].fillna(mean_age)
```

```python
np.percentile(titanic_df['Age'],25),np.percentile(titanic_df['Age'],75)
```

```python
np.percentile(titanic_df['Age'],0),titanic_df['Age'].min()
np.percentile(titanic_df['Age'],100),titanic_df['Age'].max()
np.percentile(titanic_df['Age'],50),titanic_df['Age'].median()
```

```python
sns.boxplot(x='Age',data=titanic_df)
plt.show()
```
