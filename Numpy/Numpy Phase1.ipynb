## NumPy Array and Basic


An Array is defined as a collection of items that are stored at contiguous memory locations

```python
import numpy as np
```

Create Numpy as Array using list

```python
#1D Array 
arr_id=np.array([10,20,30])
print("1D Array",arr_id)
#2D Array
arr_2d = np.array([[10,20,30],
                    [40,50,60],
                    [70,80,90]])
print("2D Array\n",arr_2d)
```

## Array Vs List

A list can't handle Mathematical operations, while Array Can Handle
An Array consume less memory than list

```python
#List Multiplication Operation
list_cost=[10,20,30,40]
print("List of sales:", list_cost*2)
#Array Multiplication Operation
arr_cost=np.array([10,20,30,40])
print("arr of sales:",arr_cost*2)
```

```python
import time as t
start=t.time()
list_mul = [i*2 for i in range(1000000)]
print("Time taken by list:",t.time()-start)
#Array
start=t.time()
arr_nums = np.arange(1000000)*2
print("Array Operation:", t.time()-start)
```

# Create array from scratch

```python
#zere array
zeros=np.zeros((3,2))
print("Zero array\n",zeros)
#One array
ones = np.ones((2,3))
print("ones array\n",ones)
#full array
full=np.full((3,3),7)
print("full array\n",full)
#random
random=np.random.random()
print("random array(0 to 1)\n",random)
#range arrange (start,stop,step)
range=np.arange(10,21,1)
print("Range 10 to 20\n",range)
```

Vector,Matrix & Tensor

```python
vector = np.array([1,2,3,4,5,6,7,8])
print("Vector: ",vector)
matrix = np.array([[1,2,3],[4,5,6]])
print("Matrix:\n ",matrix)
tensor = np.array([[[10,20],[30,40]],[[50,60],[70,90]]])
print("Tensor:\n ",tensor)
```

```python
tensor = np.array([[[10,20,100],[30,40,200]],[[50,60,300],[70,80,400]]])
print("Shape: ",tensor.shape)
print("Dimension: ",np.ndim(tensor))
print("Size: ",np.size(tensor))
print("Datatype: ",tensor.dtype)
print("Length: ",len(tensor))
print("Convert to diff data type: \n",tensor.astype(str))
```
