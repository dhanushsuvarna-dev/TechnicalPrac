String Indexing & Slicing

```python
str = "DarkChocolate"
```

```python
#indexing
print(str[2])
print(str[-len(str)])# first element using -ve index
print(str[-1])#last element of str using -ve index
```

```python
#Slicing
print(str[:4])#start_ind=0 to end-id=4 step=1
print(str[4:])#start_ind=4 to end-id=13 step=1
print(str[3::2])
rev_str = str[::-1]
print("Reverse String:",rev_str)
```

String Functions

```python
s = "he is a coder"
```

```python
#Concetination
print(s+" It's real")
#length of a String
length = len(s)
print(length)
#ends with substring
print(s.endswith("er"))
#replace old,new
print(s.replace("he is","I am"))
#captalize
print(s.capitalize())
#find 1st index of a substring
print(s.find("e"))
#count number of substring
print("count-e:",s.count("e"))
#UpperLower
print(f"Upper: {s.upper()} lower: {s.lower()}")
```
